### 2. Selling price for a desired profit after tax

$$ \text{price} = \frac{1 + \text{profit}}{1 + \text{tax}} * \text{cost} $$

```python
cost = random.randint(1000, 2000)
tax = random.randint(20, 50) / 100.0
profit = random.randint(0, 100) / 100.0

price = (1 + profit) / (1 + tax) * cost

print(f'cost {cost}, tax {tax:.0%}, profit {profit:.0%} -> price {price:.1f}')

# worked example: 1000 at 25% tax and 50% profit
assert abs((1 + 0.5) / (1 + 0.25) * 1000 - 1200.0) < 1e-9
```

### 3. The warrior and the evil wizard lizard

The trick the exercise is fishing for: **`True` is 1 and `False` is 0**, so a comparison can be multiplied straight into a formula - no `if` needed yet.

```python
sword_damage = random.randint(5, 10)
sword_damage_type = random.choice(['fire', 'ice', 'lightning', 'arcane', 'poison'])
axe_damage = random.randint(5, 10)
axe_damage_type = random.choice(['fire', 'ice', 'lightning', 'arcane', 'poison'])
evil_wizard_lizard_resist_type = random.choice(['fire', 'ice', 'lightning', 'arcane', 'poison'])
evil_wizard_lizard_damage_reduction = random.randint(0, 75) / 100.0

sword_actual_damage = sword_damage * (
    1 - evil_wizard_lizard_damage_reduction * (sword_damage_type == evil_wizard_lizard_resist_type))
axe_actual_damage = axe_damage * (
    1 - evil_wizard_lizard_damage_reduction * (axe_damage_type == evil_wizard_lizard_resist_type))

answer = sword_actual_damage > axe_actual_damage

print('Yes!' if answer else 'No!')

# the two rows of the worked example in the exercise
assert 10 * (1 - 0.25 * ('fire' == 'ice')) == 10.0
assert 10 * (1 - 0.25 * ('fire' == 'fire')) == 7.5
```

### 4. Are we there yet?

```python
distance = random.randint(30, 230) * 1.0
average_speed = random.randint(30, 90)
traffic_jam = random.randint(0, 1)
traffic_jam_penalty = random.randint(30, 60)

trip_time = distance / average_speed * 60 + traffic_jam * traffic_jam_penalty
hours = int(trip_time // 60)
minutes = int(trip_time % 60)
seconds = int(trip_time * 60 % 60)

print(f'{trip_time:.1f} minutes = {hours}h {minutes}m {seconds}s')

# 120 km at 60 km/h plus a 30 minute jam == 150 minutes == 2h 30m
assert (120.0 / 60 * 60 + 1 * 30, ) == (150.0, )
assert int(150.0 // 60) == 2 and int(150.0 % 60) == 30
```

### 5. Rock-paper-scissors

Three winning combinations, joined with `or`. Note that `and` binds tighter than `or`, so the brackets are only there for readability.

```python
your_choice = 'r'
comp_choice = random.choice('rps')

winner = ((your_choice == 'r' and comp_choice == 's')
          or (your_choice == 'p' and comp_choice == 'r')
          or (your_choice == 's' and comp_choice == 'p'))

print('YES!' if winner else 'not this time...')


def rps(you, comp):
    return ((you == 'r' and comp == 's')
            or (you == 'p' and comp == 'r')
            or (you == 's' and comp == 'p'))

assert rps('r', 's') and rps('p', 'r') and rps('s', 'p')
assert not rps('r', 'p') and not rps('r', 'r')
```

### 6. The quadratic equation

$$x_{1,2}=\dfrac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

**Careful:** with the random values the notebook generates, `b**2 - 4ac` is usually *negative*, so `** 0.5` gives a complex number. That is a good thing to point out - python does not crash, it just switches to complex numbers.

```python
a, b, c = 1, -3, 2   # a case with real roots, so the assert is readable

x_1 = (-b + (b**2 - 4*a*c) ** 0.5) / (2*a)
x_2 = (-b - (b**2 - 4*a*c) ** 0.5) / (2*a)

print(f'The solutions to {a}x^2 + {b}x + {c} = 0 are: {x_1} and {x_2}')

assert {x_1, x_2} == {2.0, 1.0}

# and what the notebook's random values normally give you:
a, b, c = 1, 1, 1
print('complex roots:', (-b + (b**2 - 4*a*c) ** 0.5) / (2*a))
```
