## 1. Objects and 2. Exceptions

The first two sections are demonstrations rather than exercises - the code in the notebook already runs. Two things worth drawing out in class:

- The cells that **deliberately fail** (`while True print(...)`, `divide(1, 0)`, `divide(3, '2')`) are the point, not a mistake. Run them, read the error, *then* fix it.
- `except:` with nothing after it catches everything, including `KeyboardInterrupt` - so a runaway loop cannot even be stopped with the stop button. That is why the notebook calls it bad practice.

````python
# the shape they should end up with
def divide(num1, num2):
    try:
        result = num1 / num2
    except ZeroDivisionError:
        print('Dividing by zero is not possible!')
        return 0
    except TypeError:
        print('Numbers can only be divided by numbers!')
        return 0
    else:
        return result
    finally:
        pass  # runs either way - closing files, releasing locks, ...


assert divide(6, 3) == 2.0
assert divide(1, 0) == 0
assert divide(3, '2') == 0
````

### 1. Seconds into years, months, days, hours, minutes, seconds

````python
def breakdown(total_seconds):
    minutes, seconds = divmod(total_seconds, 60)
    hours, minutes = divmod(minutes, 60)
    days, hours = divmod(hours, 24)
    months, days = divmod(days, 30)      # a rough month, as the exercise implies
    years, months = divmod(months, 12)
    return years, months, days, hours, minutes, seconds


def describe(total_seconds):
    years, months, days, hours, minutes, seconds = breakdown(total_seconds)
    return (f'{years} years, {months} months, {days} days, '
            f'{hours} hours, {minutes} minutes, {seconds} seconds')


print(describe(311))

# the worked example from the exercise
assert describe(311) == '0 years, 0 months, 0 days, 0 hours, 5 minutes, 11 seconds'
assert breakdown(60) == (0, 0, 0, 0, 1, 0)
assert breakdown(3600) == (0, 0, 0, 1, 0, 0)
assert breakdown(86400) == (0, 0, 1, 0, 0, 0)
````

### 2. The multiplication table of 7, starring the multiples of 3

````python
def table_of_seven(count=20):
    parts = []
    for i in range(1, count + 1):
        value = 7 * i
        parts.append(f'{value} *' if value % 3 == 0 else str(value))
    return ' '.join(parts)


print(table_of_seven())

assert table_of_seven(7).startswith('7 14 21 * 28 35 42 * 49')
# 7*i is divisible by 3 exactly when i is
assert table_of_seven(20).count('*') == len([i for i in range(1, 21) if i % 3 == 0])
````

### 3. The multiplication table of 13, only the multiples of 7

13 and 7 share no factors, so `13 * i` is divisible by 7 exactly when `i` is.

````python
def table_of_thirteen(count=50):
    return [13 * i for i in range(1, count + 1) if (13 * i) % 7 == 0]


print(table_of_thirteen())

assert table_of_thirteen() == [13 * i for i in range(1, 51) if i % 7 == 0]
assert table_of_thirteen()[:3] == [91, 182, 273]
````

### 4. Half a christmas tree, then a whole one, then a diamond

The whole point of a-b-c is that each shape is the previous one plus one idea: a height parameter, then `center()`, then two loops back to back.

````python
# 4. fixed height
for row in range(1, 8):
    print('*' * row)


# 4.a with a height parameter
def half_tree(height=7):
    return '\n'.join('*' * row for row in range(1, height + 1))


print()
print(half_tree(4))

assert half_tree(3) == '*\n**\n***'
assert len(half_tree(10).splitlines()) == 10
````

````python
# 4.b a full, centred tree
def tree(height=5):
    width = 2 * height - 1
    return '\n'.join(('*' * (2 * row - 1)).center(width)
                     for row in range(1, height + 1))


print(tree(4))

assert tree(1) == '*'
assert tree(3).splitlines() == ['  *  ', ' *** ', '*****']
assert all(len(line) == 5 for line in tree(3).splitlines())
````

````python
# 4.c a diamond - only odd sizes make sense
def diamond(size=5):
    if size % 2 == 0:
        raise ValueError(f'size must be odd, got {size}')

    half = size // 2 + 1
    top = [('*' * (2 * row - 1)).center(size) for row in range(1, half + 1)]
    # the bottom is the top without its widest row, upside down
    return '\n'.join(top + top[-2::-1])


print(diamond(3))
print()
print(diamond(5))

assert diamond(1) == '*'
assert diamond(3).splitlines() == [' * ', '***', ' * ']
assert diamond(5).splitlines() == ['  *  ', ' *** ', '*****', ' *** ', '  *  ']
try:
    diamond(4); raise AssertionError('should reject even sizes')
except ValueError as error:
    print('\n', error)
````

### 5-7. Three questions about the letter « e »

````python
# 5. does it contain an 'e'?
def has_e(text):
    return 'e' in text


# 6. szogodi dialect
def szogodi(text):
    return text.replace('e', 'ö')


# 7. how many 'e's?
def count_e(text):
    return text.count('e')


print(has_e('szeged'), szogodi('szeged'), count_e('szeged'))

assert has_e('szeged') and not has_e('paprika')
assert szogodi('szeged') == 'szögöd'
assert szogodi('Ede') == 'Edö'          # only lowercase 'e', as specified
assert count_e('szeged') == 2 and count_e('paprika') == 0
````

### 8-9. Stars between the characters, and backwards

````python
# 8. gaston -> g*a*s*t*o*n
def starify(text):
    return '*'.join(text)


# 9. live -> evil
def backwards(text):
    return text[::-1]


print(starify('gaston'), backwards('live'))

assert starify('gaston') == 'g*a*s*t*o*n'
assert starify('a') == 'a' and starify('') == ''
assert backwards('live') == 'evil'
assert backwards(backwards('anything')) == 'anything'
````

### 10. Palindromes

Part **a** is the interesting one: strip everything that is not a letter or a digit, lowercase the rest, then compare with its own reverse.

````python
def is_palindrome(word):
    return word == word[::-1]


def is_palindrome_sentence(sentence):
    cleaned = ''.join(char.lower() for char in sentence if char.isalnum())
    return cleaned == cleaned[::-1]


assert not is_palindrome('python')
assert is_palindrome('mom')

assert is_palindrome_sentence('Was it a car or a cat I saw?')
assert is_palindrome_sentence(
    'Are we not pure? “No, sir!” Panama’s moody Noriega brags. '
    '“It is garbage!” Irony dooms a man—a prisoner up to new era.')
assert not is_palindrome_sentence('Random not palindrome sentence')
print('all three sentence cases pass')
````

### 11. Interleave two lists

`zip` pairs them up, then the comprehension flattens the pairs.

````python
t1 = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
t2 = ['January', 'February', 'March', 'April', 'May', 'June',
      'July', 'August', 'September', 'Okteber', 'November', 'December']

t3 = [item for pair in zip(t2, t1) for item in pair]
print(t3[:6])

assert t3[:6] == ['January', 31, 'February', 28, 'March', 31]
assert len(t3) == len(t1) + len(t2)
````

### 12. Split into evens and odds

````python
def split_parity(numbers):
    evens = [n for n in numbers if n % 2 == 0]
    odds = [n for n in numbers if n % 2]
    return evens, odds


evens, odds = split_parity([32, 5, 12, 8, 3, 75, 2, 15])
print(evens, odds)

assert evens == [32, 12, 8, 2]
assert odds == [5, 3, 75, 15]
````

### 13. Leap years

The order of the conditions is the whole exercise: 1900 is divisible by 4 but is **not** a leap year, while 2000 is.

````python
def is_leap_year(year):
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)


assert is_leap_year(2024) and is_leap_year(2000) and is_leap_year(1600)
assert not is_leap_year(1900) and not is_leap_year(1800) and not is_leap_year(1700)
assert not is_leap_year(2023)

# cross-check against the standard library for a whole century
import calendar
assert all(is_leap_year(y) == calendar.isleap(y) for y in range(1500, 2500))
print('agrees with calendar.isleap() for 1000 years')
````

### 14-15. `generate(n, char)` and `count_char(char, string)`

````python
def generate(n, char):
    return char * n


def count_char(char, string):
    return string.count(char)


print(generate(5, '#'), count_char('e', 'easy peasy'))

assert generate(3, 'x') == 'xxx'
assert generate(0, 'x') == ''
assert count_char('e', 'easy peasy') == 2
assert count_char('z', 'easy peasy') == 0
````

### 16-17. Guess the number

Written with an injectable `_inputs` list so the solution is testable without typing. In class use plain `input()`.

Exercise **17** adds the error handling: a non-number and an out-of-range number must both be rejected *without* crashing and *without* using up a turn.

````python
def guess_a_number(from_number=1, to_number=10, _inputs=None, _target=None):
    import random

    target = _target if _target is not None else random.randint(from_number, to_number)
    queue = list(_inputs) if _inputs else None

    def ask(prompt):
        if queue is None:
            return input(prompt)
        print(prompt, end='')
        value = queue.pop(0)
        print(value)
        return value

    prompt = f'Guess a number between {from_number} and {to_number}: '
    while True:
        raw = ask(prompt)

        # exercise 17: reject rubbish instead of crashing
        try:
            guess = int(raw)
        except (TypeError, ValueError):
            prompt = "That's not a number, guess again: "
            continue

        if not from_number <= guess <= to_number:
            prompt = 'Number out of range, guess again: '
            continue

        if guess == target:
            print('Congratulations, you won!')
            return target
        prompt = 'Lower, guess again: ' if guess > target else 'Higher, guess again: '


# the exact walkthrough from the exercise text
result = guess_a_number(1, 10, _inputs=['apple', 15, 5, 3], _target=3)
assert result == 3

# 16 without the error handling still works
assert guess_a_number(1, 10, _inputs=[5, 8, 7], _target=7) == 7
print('\nboth walkthroughs reproduce exactly')
````
