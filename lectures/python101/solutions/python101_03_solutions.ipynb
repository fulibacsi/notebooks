### Exercise: Ask for a proper e-mail address

The point is the `while` loop: keep asking until the answer is acceptable. The check itself is deliberately crude - real e-mail validation is famously horrible, and that is worth one sentence in class.

```python
def looks_like_email(text):
    return text.count('@') == 1 and '.' in text.split('@')[-1]


assert looks_like_email('arthur@camelot.hu')
assert not looks_like_email('nope')
assert not looks_like_email('a@@b.hu')
assert not looks_like_email('a@bhu')

# in the notebook:
# email = input('Your e-mail address: ')
# while not looks_like_email(email):
#     email = input('That does not look like an e-mail. Try again: ')
```

### Exercise: Prisoner's dilemma

**Step 1** is one hard-coded dictionary, **step 2** replaces the index with a loop.

```python
decisions = [
    {"player1": "betray", "player2": "betray", "outcome": "Both serve 6 years"},
    {"player1": "betray", "player2": "silent", "outcome": "Player1 free, Player2 serves 10 years"},
    {"player1": "silent", "player2": "betray", "outcome": "Player1 serves 10 years, Player2 free"},
    {"player1": "silent", "player2": "silent", "outcome": "Both serve 6 months"},
]

user_choice = "betray"
ai_choice = random.choice(["betray", "silent"])

# Step 1 - just the first one
decision = decisions[0]
if decision['player1'] == user_choice and decision['player2'] == ai_choice:
    print(decision['outcome'])

# Step 2 - the same condition, inside a loop over all of them
for decision in decisions:
    if decision['player1'] == user_choice and decision['player2'] == ai_choice:
        print(decision['outcome'])


def outcome(user_choice, ai_choice):
    for decision in decisions:
        if decision['player1'] == user_choice and decision['player2'] == ai_choice:
            return decision['outcome']


assert outcome('betray', 'silent') == 'Player1 free, Player2 serves 10 years'
assert outcome('silent', 'silent') == 'Both serve 6 months'
```

### Exercise: The Seven Bridges of Königsberg

Four steps, each one a small extension of the previous one.

```python
distances = {
    'ABup': 3, 'ABdown': 4, 'ADup': 3, 'ADdown': 2,
    'AC': 10, 'BC': 7, 'CD': 6,
}
routes = [
    ['BC', 'ABdown', 'AC', 'CD', 'ADup', 'ADdown', 'ADdown', 'ABup'],
    ['ABup', 'ABdown', 'ADdown', 'ADup', 'AC', 'CD', 'ADup', 'ABup', 'BC'],
    ['ADdown', 'CD', 'BC', 'ABup', 'ABdown', 'BC', 'AC', 'ADup'],
]

# Step 1 - one bridge
route = routes[0]
bridge = route[0]
print(bridge, distances[bridge])

# Step 2 - every bridge in one route
for bridge in route:
    print(bridge, distances[bridge])

# Step 3 - add them up
route_length = 0
for bridge in route:
    route_length += distances[bridge]
print('first route:', route_length)

# Step 4 - every route
route_lengths = []
for route in routes:
    route_length = 0
    for bridge in route:
        route_length += distances[bridge]
    print(route, '->', route_length)
    route_lengths.append(route_length)

assert route_lengths == [37, 41, 42]
# ...and the one-liner version, once they have seen comprehensions:
assert route_lengths == [sum(distances[b] for b in r) for r in routes]
```

### 6. Filter the cursewords out of a text

Two details the hints are pointing at: the words in the csv are all on **one row**, and the punctuation stays glued to the words - so `strip` the commas and full stops before comparing, but keep the original word if it is clean.

```python
def clean(text, substitution='*#%=!'):
    rows = import_from_csv('./data/cursewords.csv')
    cursewords = [word.strip().lower() for row in rows for word in row]

    result = []
    for word in text.split():
        if word.strip('.,!?').lower() in cursewords:
            result.append(substitution)
        else:
            result.append(word)
    return ' '.join(result)


print(clean('A monkey and a donkey went to visit Boby .'))

assert clean('A monkey and a donkey went to visit Boby .') == \
    'A *#%=! and a *#%=! went to visit *#%=! .'
assert clean('nothing to see here') == 'nothing to see here'
```

### 7. Statistics without a t-test

Simulate 50 coin flips 10 000 times and count how often 37 or more came up heads. It is nowhere near 5%, so no - that coin is not fair.

This is worth labouring in class: they just did a hypothesis test with a `for` loop and no statistics.

```python
trials = 10000
flips = 50
threshold = 37

at_least_37 = 0
for _ in range(trials):
    heads = 0
    for _ in range(flips):
        heads += random.randint(0, 1)
    if heads >= threshold:
        at_least_37 += 1

ratio = at_least_37 / trials
print(f'{at_least_37} of {trials} runs had {threshold}+ heads ({ratio:.2%})')
print('Fair coin at 95% confidence?', 'yes' if ratio > 0.05 else 'no')

# well under 5%, so we reject the "fair coin" idea
assert ratio < 0.01
```

### 10. The Campus Party Guestlist

Part 3 is the point: a **set** gives uniqueness for free.

```python
guests = [
    {"name": "Emma",   "age": 20, "bringing": "chips"},
    {"name": "Liam",   "age": 19, "bringing": "soda"},
    {"name": "Olivia", "age": 21, "bringing": "nothing"},
    {"name": "Noah",   "age": 22, "bringing": "cake"},
    {"name": "Sophia", "age": 20, "bringing": "chips"},
]

adults = [g['name'] for g in guests if g['age'] >= 21]
chips = sum(1 for g in guests if g['bringing'] == 'chips')
unique_items = set(g['bringing'] for g in guests)

print(adults, chips, unique_items)

assert adults == ['Olivia', 'Noah']
assert chips == 2
assert unique_items == {'chips', 'soda', 'nothing', 'cake'}
```

### 11. Movie Night Voting

**Both** choices count, so every friend casts two votes. With this data everything ties on 2 - which is exactly why the exercise asks about ties.

```python
import collections

votes = [
    {'name': 'Alice',   'choice1': 'Movie A', 'choice2': 'Movie B'},
    {'name': 'Bob',     'choice1': 'Movie C', 'choice2': 'Movie A'},
    {'name': 'Charlie', 'choice1': 'Movie B', 'choice2': 'Movie C'},
]

tally = collections.Counter()
for vote in votes:
    tally[vote['choice1']] += 1
    tally[vote['choice2']] += 1

best = max(tally.values())
winners = sorted(movie for movie, count in tally.items() if count == best)

print(tally)
print('winner(s):', winners)

assert winners == ['Movie A', 'Movie B', 'Movie C']   # a three-way tie
```

### 12. Pokemon Type Counter

A pokemon can have several types, so this is a **nested** loop (or a nested comprehension).

```python
pokemon_types = {
    'Pikachu':    ['Electric'],
    'Bulbasaur':  ['Grass', 'Poison'],
    'Charmander': ['Fire'],
    'Squirtle':   ['Water'],
    'Jigglypuff': ['Normal', 'Fairy'],
    'Meowth':     ['Normal'],
    'Psyduck':    ['Water'],
    'Oddish':     ['Grass', 'Poison'],
    'Vulpix':     ['Fire'],
}
team = ['Pikachu', 'Charmander', 'Squirtle', 'Oddish']

type_counts = collections.Counter()
for pokemon in team:
    for element in pokemon_types[pokemon]:
        type_counts[element] += 1

for element, count in type_counts.items():
    print(f'{element:10s} {count}')

assert dict(type_counts) == {'Electric': 1, 'Fire': 1, 'Water': 1,
                             'Grass': 1, 'Poison': 1}
```

### 15. Lookup tables: four small filters

All four are the same two shapes - **filter** a list, or **rewrite a field** through a lookup dict. Worth saying out loud once they are done.

```python
spaceships = [
    {"name": "USS Enterprise",    "damage": 70},
    {"name": "Millennium Falcon", "damage": 40},
    {"name": "Serenity",          "damage": 85},
]
adventurers = [
    {"name": "Gandalf", "class": "Wizard"},
    {"name": "Aragorn", "class": "Ranger"},
    {"name": "Legolas", "class": "Archer"},
]
class_rename = {"Wizard": "Sorcerer", "Ranger": "Tracker", "Archer": "Sharpshooter"}
hackers = [
    {"name": "ZeroCool",      "score": 85},
    {"name": "AcidBurn",      "score": 95},
    {"name": "CrashOverride", "score": 90},
]
gear = [
    {"name": "Neon Blade",       "rarity": "rare"},
    {"name": "Optical Enhancer", "rarity": "legendary"},
    {"name": "Standard Pistol",  "rarity": "common"},
]

# a) filter
damaged = [s['name'] for s in spaceships if s['damage'] > 50]

# b) rewrite a field through the lookup dict
for adventurer in adventurers:
    adventurer['class'] = class_rename[adventurer['class']]

# c) filter again - note it is 'above 90', so 90 itself is out
top_hackers = [h['name'] for h in hackers if h['score'] > 90]

# d) filter on membership
good_gear = [g['name'] for g in gear if g['rarity'] in ('rare', 'legendary')]

print(damaged); print(adventurers); print(top_hackers); print(good_gear)

assert damaged == ['USS Enterprise', 'Serenity']
assert [a['class'] for a in adventurers] == ['Sorcerer', 'Tracker', 'Sharpshooter']
assert top_hackers == ['AcidBurn']
assert good_gear == ['Neon Blade', 'Optical Enhancer']
```

### 16. Extra challenge: Pokemon Battle Simulator

Swapping `attacker` and `defender` at the end of each turn is the trick that keeps this to one loop. Pikachu wins: 55 damage kills a 39 HP Charmander on the first hit.

```python
pokemon_stats = {
    'Pikachu':    {'HP': 35, 'Attack': 55},
    'Charmander': {'HP': 39, 'Attack': 52},
    'Squirtle':   {'HP': 44, 'Attack': 48},
    'Bulbasaur':  {'HP': 45, 'Attack': 49},
}


def battle(first, second, stats, verbose=True):
    hp = {first: stats[first]['HP'], second: stats[second]['HP']}
    attacker, defender = first, second

    while True:
        hp[defender] -= stats[attacker]['Attack']
        if verbose:
            print(f'{attacker} hits {defender} for {stats[attacker]["Attack"]} '
                  f'-> {defender} at {hp[defender]} HP')
        if hp[defender] <= 0:
            if verbose:
                print(f'{attacker} wins!')
            return attacker
        attacker, defender = defender, attacker


assert battle('Pikachu', 'Charmander', pokemon_stats) == 'Pikachu'
# whoever strikes first here wins, because every attack is a one-hit kill
assert battle('Charmander', 'Pikachu', pokemon_stats, verbose=False) == 'Charmander'
```
