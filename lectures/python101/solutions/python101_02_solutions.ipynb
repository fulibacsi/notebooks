### Exercise: Recreate *Around the World*

`capitalize()` lowercases everything after the first character, so apply it **per line**, not to the whole lyric.

```python
around_the_world = 'around the world'

line = ', '.join([around_the_world] * 4)
lyrics = '\n'.join([line.capitalize()] * 4)

print(lyrics)

assert lyrics.count('Around the world') == 4    # one per line
assert lyrics.count('around the world') == 12   # the other three per line
```

### Exercise: Decrypt the encrypted text

`encrypt(..., strength=4, level=1)` inserts `strength - 1 == 3` junk characters after every real one. So every 4th character is a real one: `[::4]`.

```python
text = 'titkositott szoveg'
encrypted = encrypt(text, strength=4, level=1)
print(encrypted)

decrypted = encrypted[::4]
print(decrypted)

assert decrypted == text
```

### Exercises: list, tuple, set, dict

```python
pi = 3.14
my_list = [1, 2, 'spam', ['eggs', 'ham'], 'spam', pi]

# access 'ham' - it is inside the nested list, so index twice
assert my_list[3][1] == 'ham'

# add ['python', '101'] to the END of my_list (append, not extend:
# extend would add the two strings separately)
my_list.append(['python', '101'])
assert my_list[-1] == ['python', '101']

A = set(['a', 'b', 'c'])
B = set(['c', 'c', 'd', 'e'])

# elements that only appear in B -> set difference
assert B - A == {'d', 'e'}
assert B.difference(A) == {'d', 'e'}
```

### Exercise: what do you think will happen and why?

**A `KeyError`.** The dictionary key is the *value* `2`, not the *variable* `a`. Rebinding `a = 6` does not reach into the dictionary and rename anything, so `dict1` still only has the key `2`. This is the difference between a name and the object it points to - worth spending a minute on.

```python
a = 2
dict1 = {a: 6}
a = 6

try:
    print(dict1[a])
except KeyError as error:
    print('KeyError:', error, '- the key is still 2')

assert 2 in dict1 and 6 not in dict1
```

### 3. Unique list from a non-unique list

```python
not_unique_list = [chr(random.randint(65, 90)) for _ in range(26)]

unique_list = list(set(not_unique_list))

print(len(not_unique_list), '->', len(unique_list))
assert len(unique_list) == len(set(not_unique_list))
assert sorted(unique_list) == sorted(set(not_unique_list))
```

### 4. Positive, negative or zero

```python
def sign(number):
    if number > 0:
        return 'positive'
    elif number < 0:
        return 'negative'
    else:
        return 'zero'


number = random.randint(-1, 1)
print(number, 'is', sign(number))

assert (sign(1), sign(-1), sign(0)) == ('positive', 'negative', 'zero')
```

### 5. FizzBuzz

The order matters: check **15 first**, otherwise 15 is caught by the `% 3` branch and never reaches `FizzBuzz`. That is the whole point of the exercise.

```python
def fizzbuzz(n):
    if n % 15 == 0:
        return 'FizzBuzz'
    elif n % 3 == 0:
        return 'Fizz'
    elif n % 5 == 0:
        return 'Buzz'
    else:
        return str(n)


numbers = list(range(30))
for n in numbers:
    print(fizzbuzz(n), end=' ')

assert [fizzbuzz(n) for n in range(1, 16)] == [
    '1', '2', 'Fizz', '4', 'Buzz', 'Fizz', '7', '8',
    'Fizz', 'Buzz', '11', 'Fizz', '13', '14', 'FizzBuzz']
```

### 6. Intolerance checking

```python
# 6.a recipes and their ingredients
ingredients = {
    'pancake':  ['flour', 'milk', 'egg', 'sugar'],
    'omelette': ['egg', 'butter', 'cheese'],
    'salad':    ['lettuce', 'tomato', 'walnut'],
}

# 6.b things that might cause a reaction
intolerancies = ['milk', 'egg', 'walnut', 'gluten']

# 6.c build the prompt string
intolerancy_string = ', '.join(intolerancies)
print('Do you have any of these intolerancies?', intolerancy_string)

assert intolerancy_string == 'milk, egg, walnut, gluten'
```

```python
# 6.d + 6.e - wrapped in a function so it is testable without input()
def can_eat(food, intolerancy):
    problem = intolerancy in ingredients.get(food, [])
    if problem:
        print(f"Careful! {food} contains {intolerancy}.")
    else:
        print(f"{food} is fine for you.")
    return not problem


assert not can_eat('pancake', 'milk')
assert can_eat('salad', 'milk')
assert not can_eat('salad', 'walnut')

# in the notebook you would read these two with input():
# intolerancy = input(f'Which of these affects you? ({intolerancy_string}) ')
# food = input(f'What would you like to eat? ({", ".join(ingredients)}) ')
# can_eat(food, intolerancy)
```
